```python
import json
from pathlib import Path
from datasets import load_dataset
```

```python
def download_ChilleD_SVAMP():
    save_dir = Path("SVAMP")
    save_dir.mkdir(exist_ok=True)
    dataset = load_dataset("ChilleD/SVAMP")

    dataset["train"].to_json(str(save_dir/"train.jsonl"))
    dataset["test"].to_json(str(save_dir/"test.jsonl"))

download_ChilleD_SVAMP()
```

```python
def download_bool_qa():
    save_dir = Path("boolq")
    save_dir.mkdir(exist_ok=True)
    dataset = load_dataset("google/boolq")

    dataset["train"].to_json(str(save_dir/"train.jsonl"))
    dataset["validation"].to_json(str(save_dir/"validation.jsonl"))

download_bool_qa()
```

```python
def download_gsm8k():
    save_dir = Path("gsm8k")
    save_dir.mkdir(exist_ok=True)
    dataset = load_dataset("openai/gsm8k", "main")

    dataset["train"].to_json(str(save_dir / "primary_train.jsonl"))
    dataset["test"].to_json(str(save_dir / "test.jsonl"))
    dataset["train"].to_json(str(save_dir / "train.jsonl"))
```

```python
def download_arc_c():
    save_dir = Path("arc-c")
    save_dir.mkdir(exist_ok=True)
    dataset = load_dataset("allenai/ai2_arc","ARC-Challenge")

    dataset["train"].to_json(str(save_dir/"train.jsonl"))
    dataset["validation"].to_json(str(save_dir/"validation.jsonl"))
    dataset["test"].to_json(str(save_dir/"test.jsonl"))
```

```python
def download_arc_e():
    save_dir = Path("arc-e")
    save_dir.mkdir(exist_ok=True)
    dataset = load_dataset("allenai/ai2_arc","ARC-Easy")

    dataset["train"].to_json(str(save_dir/"train.jsonl"))
    dataset["validation"].to_json(str(save_dir/"validation.jsonl"))
    dataset["test"].to_json(str(save_dir/"test.jsonl"))
```

```python
def download_cola():
    save_dir = Path("cola")
    save_dir.mkdir(exist_ok=True)
    dataset = load_dataset("nyu-mll/glue","cola")


    dataset["train"].to_json(str(save_dir/"train.jsonl"))
    dataset["validation"].to_json(str(save_dir/"validation.jsonl"))
    dataset["test"].to_json(str(save_dir/"test.jsonl"))
```

```python
def download_sst2():
    save_dir = Path("sst2")
    save_dir.mkdir(exist_ok=True)
    dataset = load_dataset("stanfordnlp/sst2")


    dataset["train"].to_json(str(save_dir/"train.jsonl"))
    dataset["validation"].to_json(str(save_dir/"validation.jsonl"))
    dataset["test"].to_json(str(save_dir/"test.jsonl"))
```

```python
download_arc_c()
download_arc_e()
download_cola()
download_gsm8k()
download_sst2()
```
